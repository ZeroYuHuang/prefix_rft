```python
import os
import uuid
from datasets import load_dataset

if __name__ == "__main__":
    data_file = "/home/ma-user/work/dev/_huggingface/datasets/_data/raw_dataset/math/train.jsonl"
    from datasets import load_dataset

    # 直接指向具体的 arrow 文件
    dataset = load_dataset(
        "json", 
        data_files=data_file, 
        split="train"
    )

    print(dataset)
```

```python
dataset[0]
```

```python
def process(example):
    raw_prompt = example["problem"]
    example["prompt"] = [
        {"content": "Please reason step by step and put your final answer in \\boxed{}", "role": "system"},
        {"content": raw_prompt, "role": "user"},
    ]
    example["response"] = example["solution"]
    return example
dataset = dataset.map(process, num_proc=8)
print(dataset[0])
```

```python
HOME_DIR = "/home/ma-user/work/hzy-dev"
save_dir = f"{HOME_DIR}/_data/processed_dataset_new/math_sft"
dataset.to_parquet(os.path.join(save_dir, "train.parquet"))
```
