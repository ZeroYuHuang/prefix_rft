```python
Qwen3_8B_Base_chat_template =   """
{%- for message in messages %}\n    
    {%- if (message.role == "user") or (message.role == "system" and loop.first) %}\n       
        {{- \'<|im_start|>\' + message.role + \'\\n\' + message.content + \'<|im_end|>\' + \'\\n\' }}\n    
    {%- elif message.role == "assistant" %}\n        
        {%- set content = message.content %}\n              
        {{- \'<|im_start|>\' + message.role + \'\\n\' }}\n          
        {{- content.lstrip(\'\\n\') }}\n            
        {%- if not loop.last %}\n            
            {{- \'<|im_end|>\\n\' }}\n        
        {%- endif %}\n    
    {%- endif %}\n
{%- endfor %}\n
{%- if add_generation_prompt %}\n    
    {{- \'<|im_start|>assistant\' + \'\\n\' }}\n    
{%- endif %}
"""
tok.chat_template = Qwen3_8B_Base_chat_template
print(tok.apply_chat_template(prompt_continue, tokenize=False, continue_final_message=True))
print()
print(tok.apply_chat_template(prompt, tokenize=False, add_generation_prompt=True))
```
